Goal: fill missing numeric and categorical values, create a new feature, and verify that no missing values remain.

## 1. Create the sample dataset

```python
import pandas as pd
import numpy as np

data = {
    "age": [22, 35, np.nan, 29, 52, 46, 31, np.nan, 60, 38],
    "monthly_fee": [45, 80, 95, np.nan, 110, 100, 72, 55, 120, 85],
    "months_active": [3, 24, 36, 8, 48, 30, 15, 6, 60, 20],
    "support_tickets": [4, 1, 0, 3, 0, 1, 2, 4, 0, 1],
    "plan_type": [
        "Basic", "Premium", "Premium", "Basic", None,
        "Premium", "Standard", "Basic", "Premium", None
    ],
    "churned": [1, 0, 0, 1, 0, 0, 1, 1, 0, 0]
}

df = pd.DataFrame(data)
df.head()
```

## 2. Check for missing values

```python
df.isnull().sum()
```

## 3. Fill numeric missing values using the median

Learning note: I first placed `.median()` after `fillna()`, which assigned a single median value to the entire age column. The correct order is to calculate the median first, then fill only missing entries. In Jupyter, I had to reset `df` after overwriting the column.

```python
age_median = df["age"].median()
print(age_median)
df["age"] = df["age"].fillna(age_median)

fee_median = df["monthly_fee"].median()
print(fee_median)
df["monthly_fee"] = df["monthly_fee"].fillna(fee_median)

df
```

## 4. Fill the missing plan types using the mode

Learning note: `.mode()` returns a Series, so `[0]` selects the first most common category for `fillna()`.

```python
plan_mode = df["plan_type"].mode()[0]
print(plan_mode)
df["plan_type"] = df["plan_type"].fillna(plan_mode)
df
```

## 5. Add the `tickets_per_month` feature

```python
df["tickets_per_month"] = df["support_tickets"] / df["months_active"]
df
```

## 6. Verify that there are no missing values left

```python
df.isnull().sum()
```
