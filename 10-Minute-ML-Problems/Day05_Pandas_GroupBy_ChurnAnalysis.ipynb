```python
import pandas as pd

customers = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "plan": ["Basic", "Premium", "Basic", "Premium","Basic", "Standard", "Standard", "Premium"],
    "monthly_fee": [40, 90, 45, 85, 50, 65, 60, 95],
    "support_tickets": [4, 1, 5, 2, 3, 2, 4, 0],
    "churned": [1, 0, 1, 0, 0, 0, 1, 0]
})

print(customers)
```

The average monthly fee across all customers.


```python
monthlyfee_avg = customers["monthly_fee"].mean()
print(f"Average Monthly Fee: {monthlyfee_avg}")
```

The average monthly fee separated by subscription plan.


```python
fee_seperated = customers.groupby("plan")["monthly_fee"].mean()
print(f"Average Monthly Fee by Plan:\n{fee_seperated}")
```

The average number of support tickets across all customers.


```python
support_avg = customers['support_tickets'].mean()
print(f"Average Support Tickets: {support_avg}")
```

The average number of support tickets separated by subscription plan.


```python
support_seperated = customers.groupby("plan")["support_tickets"].mean()
print(f"Average Support Tickets by Plan:\n{support_seperated}") 
```

Calculate the overall churn rate in two ways: count customers who churned and divide by the total, or use the mean of a column containing only zeros and ones.


```python
churned_count = customers['churned'].sum()

# count all customers
total_customers = customers["churned"].count()

#calculate churn rate
churn_rate = churned_count / total_customers
print (f"Churn Rate: {churn_rate:.2%}")
churn_avg = customers['churned'].mean()
print(f"Average Churn Rate: {churn_avg:.2%}")

```

Calculate the churn rate for each subscription plan. The mean of the `churned` column gives the proportion of customers who left.


```python
churned_plan = customers.groupby("plan")["churned"].mean()
print(f"Churn Rate by Plan:\n{churned_plan.map('{:.2%}'.format)}")
```
