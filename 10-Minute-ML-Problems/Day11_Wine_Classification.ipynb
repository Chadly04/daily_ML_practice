## How KNN works 
KNN predicts by looking at similar examples it has already seen.
Imagine you give it a new wine sample:
1. It compares that wine’s measurements—alcohol, acidity, and the other features—with the training wines.
2. It finds the closest matches. “Closest” means their measurements are most similar.
3. Those wines vote on the new wine’s class.
K is how many wines get to vote.
Suppose K=5, and the five closest training wines have these labels:
Class 1
Class 1
Class 2
Class 1
Class 0

Class 1 receives three votes, so KNN predicts class 1.
Your scaler helps make the comparison fair: a feature measured in hundreds shouldn’t dominate just because another feature is measured in decimals.

### Your Task 2 code


````python
#1
X = df.drop("target", axis= 1 )
#2
y= df["target"]
#3
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test =train_test_split(X,y, random_state = 42,test_size=0.20, stratify=y)
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
````

## Task 3 — Build the classification pipeline

Create a pipeline containing:

- `StandardScaler()`
- `KNeighborsClassifier(n_neighbors=5)`

Fit the pipeline using only `X_train` and `y_train`.

**Think about:** Why might scaling be especially important for KNN?

### Your Task 3 code


````python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

model_k = Pipeline([
    ("scaler", StandardScaler()),
    ("Kneighbor", KNeighborsClassifier())
    
])
````

````python
model_k.fit(X_train,y_train)
````

Scaling for Kneighbors is important beacuse it goes off of distance 

````python
y_pred = model_k.predict(X_test)
````

## Task 4 — Evaluate the model

Use the trained pipeline to predict `X_test`, then calculate:

- Accuracy
- Precision
- Recall
- F1 score
- Confusion matrix

Because this problem has more than two classes, use `average="macro"` for precision, recall, and F1.

Display each result with a clear label.

### Your Task 4 code


````python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

accuracy = accuracy_score(y_test, y_pred)
print("accuracy: ",accuracy)

precision = precision_score(y_test, y_pred, average="macro")
print("precision: ",precision)

recall = recall_score(y_test, y_pred, average="macro")
print("recall: ",recall)

f1= f1_score(y_test, y_pred, average="macro")
print("f1 Score: ",f1)

confusion = confusion_matrix(y_test, y_pred)
print(confusion)
````

## Concept questions

### Question 1
Why is this a classification problem instead of a regression problem?

**My answer:** We are predicting which wine class a sample belongs to, rather than a numerical amount. 

### Question 2
Why does KNN usually need scaled features?

**My answer:** KNN uses distance to find similar samples. Scaling prevents features with larger numerical ranges from dominating that distance.

### Question 3
What could happen if `K` is too small, such as `K=1`?

**My answer:** Only the closest sample votes. If that sample is unusual or mislabeled, the model could make an incorrect prediction.

### Question 4
What could happen if `K` is too large?

**My answer:** 1. Too many distant, less similar samples get to vote. This can hide useful patterns and make predictions less accurate.
