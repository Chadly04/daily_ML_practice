```python
import numpy as np

# columns:
# [months_active, support_tickets, monthly_fee]

customers = np.array([
    [12, 3, 70],
    [24, 2, 85],
    [6,  4, 55],
    [36, 1, 95],
    [3,  3, 45],
    [18, 0, 80]
])

```

tickets_per_month = support_tickets / months_active

```python
support_tickets = customers[:, 1]  # all rows, column 1
months_active = customers[:, 0]  # all rows, column 0
tickets_per_month = support_tickets / months_active
print(tickets_per_month)
```

estimated_total_spent = monthly_fee * months_active

```python
months_active = customers[:, 0]  # all rows, column 0
monthly_fee = customers[:,2]  # all rows, column 2
estimated_total_spent = monthly_fee * months_active
print(estimated_total_spent)
```

Then combine those two new features with the original customers array so that the final array has 5 columns: months_active | support_tickets | monthly_fee | tickets_per_month | estimated_total_spent

## Combine features into the original customer array

**Mistake and correction:** I initially tried `customers.append(tickets_per_month, estimated_total_spent, axis=1)`, but NumPy arrays do not provide a pandas-style `.append()` method. `np.column_stack()` creates a new array with the original data and both new features placed side by side. `customers` stays unchanged.

```python
# np.column_stack() adds arrays side by side as columns.
# The original customers array has 6 rows and 3 columns.
# Add tickets_per_month and estimated_total_spent as 2 new columns.
# The outer () call the function; the inner () group the arrays.
# Unlike pandas, NumPy arrays don't have an .append() method.
final_data = np.column_stack((
    customers,
    tickets_per_month,
    estimated_total_spent
))

# This should have 6 customers and 5 features.
print(final_data)
print(final_data.shape)
```

## Check the results

The expected final shape is `(6, 5)`: six customers, five features. The final two columns are ticket frequency and estimated spending. Values such as `2.50000000e-01` are scientific notation for `0.25`; `np.set_printoptions(suppress=True, precision=2)` makes the display easier to read.

```python
# Optional: display decimal numbers instead of scientific notation
np.set_printoptions(suppress=True, precision=2)
print(final_data)
```

## Concept question and what I learned

**Question:** Why might `tickets_per_month` help predict customer churn more than the raw number of support tickets?

**My answer (from our discussion):** Customers having problems more frequently may be more likely to leave. Two customers can have the same total tickets but very different ticket rates if one has been active much longer. This feature captures how often issues occur instead of just the total number of tickets. It's a possible signal, not proof that a customer will churn.
